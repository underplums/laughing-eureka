## Block Coverage Report

Запускать после того, как в основном ноутбуке уже есть `df`.

```python
import pandas as pd

block_report = []

for block_name, block_features in feature_blocks.items():
    present = [f for f in block_features if f in df.columns]
    missing = [f for f in block_features if f not in df.columns]

    block_report.append({
        "block": block_name,
        "expected_features": len(block_features),
        "present_features": len(present),
        "missing_features": len(missing),
        "missing": missing,
        "mean_missing_share": df[present].isna().mean().mean() if present else None,
        "max_missing_share": df[present].isna().mean().max() if present else None,
        "mean_zero_share": (df[present] == 0).mean().mean() if present else None,
    })

block_report_df = pd.DataFrame(block_report).sort_values("missing_features", ascending=False)
block_report_df
```

## Block Coverage Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns

plot_df = block_report_df.sort_values("present_features", ascending=False)

plt.figure(figsize=(14, 7))
sns.barplot(data=plot_df, x="block", y="present_features", color="#4C78A8")
plt.xticks(rotation=45, ha="right")
plt.title("Loaded features by logical block")
plt.xlabel("")
plt.ylabel("present features")
plt.tight_layout()
plt.show()
```
